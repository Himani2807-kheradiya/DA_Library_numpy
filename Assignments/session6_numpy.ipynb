TASK-4.
Create a NumPy array of the first 20 natural numbers. Use step slicing to print every 3rd number starting from the second element.<br><br><em><strong>Hint:</strong> Use the slice notation with a step value.</em>

```python
natural_numbers = np.arange(1, 21)
```

```python
natural_numbers
```

```python
every_third = natural_numbers[1::3]
```

```python
every_third
```

```python
print("4. Every 3rd Number Starting from Second Element:")
print(every_third)
```

TASK-5.
Given a NumPy array of Flipkart product prices, use boolean indexing to extract all prices greater than 500. Print the resulting array.

```python
flipkart_prices = np.array([250, 650, 450, 800, 350, 1200, 500, 750])
```

```python
flipkart_prices
```

```python
expensive_products = flipkart_prices[flipkart_prices > 500]
```

```python

print("5. Prices Greater Than 500:")
print(expensive_products)
```

TASK-6.
You have an array of IPL team scores: np.array([210, 180, 195, 220, 205, 175]). Use np.where() to find the indices of all scores above 200 and print these indices.

```python

ipl_scores = np.array([210, 180, 195, 220, 205, 175])
```

```python
ipl_scores
```

```python
indices = np.where(ipl_scores > 200)
```

```python

print("\n6. Indices of Scores Above 200:")
print(indices[0])
```
