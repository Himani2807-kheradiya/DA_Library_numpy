Given a 2D NumPy array of daily step counts for 5 days (each row is a day, columns are morning and evening), use reshape() to convert it into a 1D array, then back to a 2D array with 5 rows and 2 columns.<br><br><em><strong>Hint:</strong> Use the shape attribute to check your array after each reshape.</em>

```python

steps = n = np.random.randint(10,100,10).reshape(5,2)
```

```python
steps
```

```python
steps = np.array([
    [3500, 4200],   
    [4000, 5000],  
    [3200, 4500],   
    [5000, 5500],   
    [4500, 4800] 
])

print("\n2. Original 2D Steps Array:")
print(steps)
print("Shape:", steps.shape)
```

```python
# Convert 2D array into 1D
steps_1d = steps.reshape(-1)

print("After converting to 1D:")
print(steps_1d)
print("Shape:", steps_1d.shape)
```

```python

# Convert 1D array back to 2D with 5 rows and 2 columns
steps_2d = steps_1d.reshape(5, 2)

print("After converting back to 2D:")
print(steps_2d)
print("Shape:", steps_2d.shape)
```

TASK-3.
Build a NumPy array representing the prices of 12 food items from a Zomato order, then use ravel(), flatten(), and resize() to create different shaped versions of the data and print each result.<br><br><em><strong>Constraint:</strong> Show the difference between ravel() and flatten() in your code comments.</em>

```python

prices = np.array([
    120, 150, 80, 200,
    99, 180, 250, 75,
    130, 160, 90, 220
])

print("\n3. Original Food Prices:")
print(prices)
print("Shape:", prices.shape)
```

```python
prices_2d = prices.reshape(3, 4)

print(prices_2d)
```

```python
prices_ravel = prices_2d.ravel()

print(prices_ravel)
```

```python
prices_flatten = prices_2d.flatten()

print(prices_flatten)
```

```python
prices_resize = prices_2d.copy()
prices_resize.resize(4, 3)
```

```python
print("Using resize() - 4x3:")
print(prices_resize)
print("Shape:", prices_resize.shape)
```

TASK-5.
Given a 1D NumPy array of 15 Flipkart product ratings, use reshape() to convert it into a 3x5 array, then use flatten() to return it to a 1D array. Explain in a comment when you would use flatten() versus ravel() in real projects.

```python
ratings = np.array([
    5, 4, 3, 5, 4,
    2, 5, 3, 4, 5,
    4, 3, 5, 2, 4
])
```

```python
ratings
```

```python
ratings.shape
```

```python
# Convert 1D array into 3x5
ratings_2d = ratings.reshape(3, 5)
```

```python
ratings_2d
```

```python
print("Shape:", ratings_2d.shape)
```

```python
ratings_1d = ratings_2d.flatten()

print("After flatten():")
print(ratings_1d)
print("Shape:", ratings_1d.shape)
```

```python
#If you only need to read/process the flattened data and memory efficiency matters, ravel() is often useful. If you need an independent array that can be modified safely without affecting the original, use flatten().
```
