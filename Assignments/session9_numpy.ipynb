```python
import numpy as np
```

TASK-1.
Create a NumPy array named prices with these values: [299, 499, 799, 0, 1599, -1, 899]. Print the array and its data type.

```python
prices = np.array([299, 499, 799, 0, 1599, -1, 899])
```

```python
print("1. Original Prices:")
print(prices)
```

```python
print("Data Type:")
print(prices.dtype)
```

TASK-2.
Some entries in the prices array are invalid (0 or negative). Replace all values less than or equal to zero with the average of the remaining positive prices.<br><br><em><strong>Hint:</strong> Use boolean indexing and the mean() function.</em>

```python
positive_prices = prices[prices > 0]
```

```python
average_price = np.mean(positive_prices)
```

```python
print(" Average of Positive Prices:")
print(average_price)
```

```python
cleaned_prices = prices.copy()
cleaned_prices[cleaned_prices <= 0] = average_price
```

```python
print("Cleaned Prices:")
print(cleaned_prices)
```

TASK-3.
Suppose you have a NumPy array quantities = [2, 1, 3, 4, 2, 1, 5]. Calculate the total bill for each item by multiplying the cleaned prices array with quantities, and print the resulting array.

```python
quantities = np.array([2, 1, 3, 4, 2, 1, 5])
```

```python
item_bills = cleaned_prices * quantities
```

```python
print("\n3. Bill for Each Item:")
print(item_bills)
```

TASK-4.
Generate and print a summary report: show the minimum, maximum, average, and total of the cleaned prices array, and also the total bill for all items combined.<br><br><em><strong>Hint:</strong> Use NumPy functions like min(), max(), mean(), and sum().</em>

```python
minimum_price = np.min(cleaned_prices)
maximum_price = np.max(cleaned_prices)
average_cleaned_price = np.mean(cleaned_prices)
total_prices = np.sum(cleaned_prices)
```

```python
total_bill = np.sum(item_bills)
```

```python
print(" Summary Report")
print("-----------------------")
print("Minimum Price:", minimum_price)
print("Maximum Price:", maximum_price)
print("Average Price:", average_cleaned_price)
print("Total of Cleaned Prices:", total_prices)
print("Total Bill:", total_bill)
```
