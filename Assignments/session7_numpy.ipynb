TASK-4.
Simulate a WhatsApp group chat: create a 2D NumPy array where each row is a user and each column is the number of messages sent per day for a week. Use np.insert() to add a new user (row) with their message counts, then use np.delete() to remove the user who sent the least messages overall.

```python
messages = np.array([
    [10, 15, 20, 12, 18, 14, 16],  
    [25, 20, 30, 22, 28, 24, 26],  
    [5, 8, 6, 7, 9, 4, 6],         
    [18, 15, 20, 17, 19, 21, 16]    
])
```

```python
messages
```

```python
print(" Original WhatsApp Messages:")
print(messages)
```

```python
new_user = np.array([[12, 10, 15, 11, 14, 13, 16]])
```

```python
messages = np.insert(messages, messages.shape[0], new_user, axis=0)
```

```python
print("After Adding New User:")
print(messages)
```

```python
total_messages = np.sum(messages, axis=1)
```

```python
print("Total Messages by Each User:")
print(total_messages)
```

```python
least_user_index = np.argmin(total_messages)
```

```python
print("User with Least Messages: User", least_user_index + 1)
```

```python
messages = np.delete(messages, least_user_index, axis=0)
```

```python

print("After Removing User with Least Messages:")
print(messages)
```

TASK-5.
Given a NumPy array of YouTube video view counts, use .view() to create a view and .copy() to create a copy. Modify the first element in each and print all arrays to demonstrate the difference between view and copy.<br><br><em><strong>Hint:</strong> Observe which changes affect the original array.</em>

```python
views = np.array([1000, 2000, 3000, 4000, 5000])
```

```python
views_view = views.view()
views_copy = views.copy()
```

```python
views_view[0] = 9999
```

```python
views_copy[0] = 8888
```

```python

print("5. view() vs copy()")

print("Original Array:")
print(views)

print("View:")
print(views_view)

print("Copy:")
print(views_copy)
```
